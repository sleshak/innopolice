```python
import random

def random_values(a: int, b: int, length: int) -> list:
    return [random.randint(a,b) for _ in range(length)]

test1 = random_values(10,100,5)
print(test1)
```

```python
def distinct_values(lst: list) -> dict:
    dist = {}
    for i in lst: 
        dist[i] = dist.get(i, 0) + 1
    return dist

test2 = distinct_values([1,1,1,1,2,4,647,234,2,657,2,34,46])
print(test2)
```

```python
def a_lot_of_calculations(lst: list) -> dict:

    minimum = min(lst)

    maximum = max(lst)

    average = sum(lst) / len(lst)

    sorted_lst = sorted(lst)
    mid = len(lst) // 2
    if len(lst) % 2 == 1:
        median = sorted_lst[mid]
    else:
        median = (sorted_lst[mid - 1] + sorted_lst[mid]) / 2
    
    freq = {}
    for i in lst:
        if i in freq:
            freq[i] += 1
        else:
            freq[i] = 1
    
    max_count = max(freq.values())
    modes = [num for num, count in freq.items() if count == max_count]

    if max_count == 1:
        mode = 0
    else:
        mode = modes
        

    return {
        'min': minimum,
        'max': maximum,
        'average': average,
        'median': median,
        'mode': mode
    }

test3 = a_lot_of_calculations([1, 2, 3, 1, 2, 2, 4, 5, 1])
print(test3)
```
